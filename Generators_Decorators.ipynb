```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
    countdown(3)
for i in countdown(5):
    print(i)
    
```

```python
def logger(func):
    def wrapper():
        print("Function Started")
        func()
        print("Function Ended")
    return wrapper
@logger
def greet():
    print("Emma health baundhaa")
greet()
```

# Decorators

```python
# Decorator Example to find execution time
import time
def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Time Taken:", end-start)
        return result
    return wrapper

@timer
def process_data():
          for i in range(1000000):
                    pass
process_data()
```

```python
# Example 2: Student Authentication System - Only authenticated students should access dashboard
is_logged_in = True
def login_required(func):
          def wrapper():
                    if is_logged_in:
                              func()
                    else:
                              print("Please Login First")
          return wrapper

@login_required
def dashboard():
    print("Welcome To Student Dashboard")
dashboard()
is_logged_in = False # Change
dashboard()
```
